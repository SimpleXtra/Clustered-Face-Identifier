```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("../outputs/logs/train4-results.csv")
df.head()
```

```python
plt.title("Box Loss")
plt.plot(df["train/box_loss"], label="Training")
plt.plot(df["val/box_loss"], label="Validation")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
plt.title("Classification Loss")
plt.plot(df["train/cls_loss"], label="Training")
plt.plot(df["val/cls_loss"], label="Validation")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
plt.title("Distribution Focal Loss")
plt.plot(df["train/dfl_loss"], label="Training")
plt.plot(df["val/dfl_loss"], label="Validation")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```
